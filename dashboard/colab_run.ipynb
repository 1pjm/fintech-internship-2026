```python
import getpass, os

for name in ["SERVICE_KEY_APT", "SERVICE_KEY_RH", "SERVICE_KEY_SH", "SERVICE_KEY_OFFI"]:
    val = getpass.getpass(f"{name} (건너뛰려면 Enter): ")
    if val.strip():
        os.environ[name] = val.strip()
```
